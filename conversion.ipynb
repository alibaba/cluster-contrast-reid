ID 2, camera 1-sequence 1, frame 451, third detected box <br>

| Part    | Example | Meaning                                                                                                                             |
|---------|---------|-------------------------------------------------------------------------------------------------------------------------------------|
| **PID** | `0002`  | Person ID 2 (labels `0001`–`1501`). `0000` is reserved for distractor/background images in **bounding_box_test**.                    |
| **c#**  | `c1`    | Camera 1 (the dataset was captured with six cameras).                                                                               |
| **s#**  | `s1`    | First video clip/sequence from that camera.                                                                                         |
| **FFFFFF** | `000451` | Frame index 451 in that clip (videos were 25 fps).                                                                              |
| **BB**  | `03`    | Bounding-box ID assigned by the DPM detector on that frame. `00` appears only in **gt_bbox** / **gt_query** folders and denotes a hand-labeled box. |
